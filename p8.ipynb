```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('sink.jpg')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
low=100
high=200
```

```python
without_background = np.zeros_like(gray)
# without_background[(gray > low) & (gray < high)] = 255
```

```python
height, width = gray.shape

for i in range(height):
    for j in range(width):
        if gray[i, j] > low and gray[i, j] < high:
            without_background[i, j] = 255
```

```python
with_background = np.copy(gray)
# with_background[(gray > low) & (gray < high)] = 255
```

```python
for i in range(height):
    for j in range(width):
        if gray[i, j] > low and gray[i, j] < high:
            with_background[i, j] = 255
```

```python
plt.figure(figsize=(10, 6))
plt.subplot(1, 3, 1)
plt.imshow(gray, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(without_background, cmap='gray')
plt.title('Without Background')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(with_background, cmap='gray')
plt.title('With Background')
plt.axis('off')
plt.tight_layout()  
plt.show()
```
