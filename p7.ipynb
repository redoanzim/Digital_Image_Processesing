```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('andrew.jpg')
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
Negative =255-image_rgb
```

```python
_,binary = cv2.threshold(
    image_rgb,
    127,
    255,
    cv2.THRESH_BINARY
)
```

```python
plt.figure(figsize=(10, 6))
plt.subplot(1, 3, 1)
plt.imshow(image_rgb)
plt.title('Original')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(Negative)
plt.title('Negative')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(binary)
plt.title('Binary')
plt.axis('off')

plt.tight_layout()
plt.show()
```
