```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("andrew.jpg", 0)
```

```python
plt.figure(figsize=(12, 8))

for i in range(8):
    bit_plane = (img >> i) & 1

    plt.subplot(2, 4, i + 1)
    plt.imshow(bit_plane, cmap="gray")
    plt.title("Bit Plane " + str(i))
    plt.axis("off")

plt.show()
```
