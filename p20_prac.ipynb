```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image =cv2.imread("andrew.jpg", 0)
plt.imshow(image, cmap='gray')
```

```python
_, binary = cv2.threshold(image,127,255,cv2.THRESH_BINARY)
plt.imshow(binary, cmap='gray')
```

```python
num_labels, labels =cv2.connectedComponents(binary)

segemented_img = np.uint8(255*labels/max(num_labels-1,1))
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(binary, cmap='gray')
plt.title('Binary Image')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(segemented_img, cmap='nipy_spectral')
plt.title('Segmented Image')
plt.axis('off')

plt.show()

print("total number of segments: ", num_labels - 1)
```
