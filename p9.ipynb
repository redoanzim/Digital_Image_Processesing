```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('andrew.jpg')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
plt.imshow(gray, cmap='gray')
plt.axis('off')
plt.show()
```

```python
#contast stretching
r_min =np.min(gray)
r_max = np.max(gray)
contrast_stretched = ((gray - r_min) / (r_max - r_min) * 255).astype(np.uint8)
```

```python
plt.imshow(contrast_stretched, cmap='gray')
plt.axis('off')
plt.show()
```

```python
#bit slicing 
plt.figure(figsize=(12, 8))

for i in range(8):
    bit_plane = (gray >> i) & 1

    plt.subplot(2, 4, i + 1)
    plt.imshow(bit_plane, cmap="gray")
    plt.title("Bit Plane " + str(i))
    plt.axis("off")

plt.show()
```
