```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
orginal =cv2.imread('sink.jpg',cv2.IMREAD_GRAYSCALE)
```

```python
plt.imshow(orginal,cmap='gray')
```

```python
orginal.shape
```

```python
#changing spatial resulation of the image
spatial_256 = cv2.resize(orginal,(256,256))
spatial_128 = cv2.resize(orginal,(128,128))
spatial_64 = cv2.resize(orginal,(64,64))
```

```python
#changing intensity resulation of the image
# step_size = 256//level
intensity_256 = (orginal//1)*1
intensity_16 = (orginal//16)*16
intensity_4 = (orginal//64)*64
```

```python
# -----------------------------
# Display Spatial Resolution
# -----------------------------

plt.figure(figsize=(10, 6))

plt.subplot(2, 2, 1)
plt.imshow(orginal, cmap="gray")
plt.title("Original")
plt.axis("off")


plt.subplot(2, 2, 2)
plt.imshow(spatial_256, cmap="gray")
plt.title("256 x 256")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(spatial_128, cmap="gray")
plt.title("128 x 128")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(spatial_64, cmap="gray")
plt.title("64 x 64")
plt.axis("off")


plt.tight_layout()
plt.show()
```

```python
# -----------------------------
# Display Intensity Resolution
# -----------------------------

plt.figure(figsize=(10, 6))

plt.subplot(1, 3, 1)
plt.imshow(intensity_256, cmap="gray")
plt.title("256 Levels")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(intensity_16, cmap="gray")
plt.title("16 Levels")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(intensity_4, cmap="gray")
plt.title("4 Levels")
plt.axis("off")

plt.tight_layout()
plt.show()
```
