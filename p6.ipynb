```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
image_1 = cv2.imread('sink.jpg')
image_1 = cv2.cvtColor(image_1, cv2.COLOR_BGR2RGB)
image_2 = cv2.imread('andrew.jpg')
image_2 = cv2.cvtColor(image_2, cv2.COLOR_BGR2RGB)
```

```python
#image resize

image_1 = cv2.resize(image_1,(256,256))
image_2 = cv2.resize(image_2,(256,256))
```

```python
#arithmetic operations
addition = cv2.add(image_1,image_2)
subtraction = cv2.subtract(image_1,image_2)
And = cv2.bitwise_and(image_1,image_2)
Or = cv2.bitwise_or(image_1,image_2)
```

```python
plt.figure(figsize=(10, 6))
plt.subplot(2,3,1)
plt.imshow(image_1)
plt.title('Image 1')
plt.axis('off')

plt.subplot(2,3,2)
plt.imshow(image_2)
plt.title('Image 2')
plt.axis('off')

plt.subplot(2,3,3)
plt.imshow(addition)
plt.title('Addition')
plt.axis('off')

plt.subplot(2,3,4)
plt.imshow(subtraction)
plt.title('Subtraction')
plt.axis('off')

plt.subplot(2,3,5)
plt.imshow(And)
plt.title('AND')
plt.axis('off')

plt.subplot(2,3,6)
plt.imshow(Or)
plt.title('OR')
plt.axis('off')

plt.tight_layout()
plt.show()
```
