```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
image = cv2.imread('sink.jpg')
image = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(image)
```

```python
image.shape
```

```python
height, width = image.shape[:2]
```

```python
center = (width // 2, height // 2)
```

```python
#rotate the image by 45 degrees
rotation_matrix = cv2.getRotationMatrix2D(
    center,
    90,
    1
)
rotated_image = cv2.warpAffine(
    image,
    rotation_matrix,
    (width, height)
)
```

```python
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image)   
plt.title("Original Image")

plt.subplot(1, 2, 2)
plt.imshow(rotated_image)
plt.title("Rotated Image")


plt.show()
```

```python
#translate the image by 100 pixels to the right and 50 pixels down
# -----------------------------
# 3. Translation
# -----------------------------
translation_matrix = cv2.getRotationMatrix2D(
    (0, 0),
    0,
    1
)

translation_matrix[0, 2] = 100  # Move right
translation_matrix[1, 2] = 40   # Move down

translated = cv2.warpAffine(
    image,
    translation_matrix,
    (width, height)
)
```

```python
# -----------------------------
# Display Results
# -----------------------------
plt.figure(figsize=(12, 8))

plt.subplot(1, 3, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")


plt.subplot(1, 3 ,2)
plt.imshow(rotated_image)
plt.title("Rotated 90 Degrees")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(translated)
plt.title("Translated")
plt.axis("off")

plt.tight_layout()
plt.show()
```
