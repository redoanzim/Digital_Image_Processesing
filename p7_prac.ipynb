```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('sink.jpg',0)
image = cv2.cvtColor(image,cv2.COLOR_BGR2RGB) 
plt.imshow(image)
plt.show()
```

```python
Negative = 255-image
plt.imshow(Negative)    
plt.show()
```

```python
_,binary= cv2.threshold(
    image,
    127,
    255,
    cv2.THRESH_BINARY
)
```

```python
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.imshow(image)
plt.axis('off')
plt.title('Original Image')

plt.subplot(1,3,2)
plt.imshow(Negative)
plt.axis('off')
plt.title('Negative Image')

plt.subplot(1,3,3)
plt.imshow(binary)
plt.axis('off')
plt.title('Binary Image')

plt.show()
```
