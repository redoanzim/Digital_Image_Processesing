```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image =cv2.imread('sink.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

plt.imshow(image, cmap='gray')
plt.axis('off')
```

```python
min = np.min(image)
max = np.max(image)

contrast_stretched = np.uint8((image-min)/(max-min)*255)
```

```python
plt.imshow(contrast_stretched, cmap='gray')
plt.axis('off')
plt.show()
```

```python

plt.figure(figsize=(12,6))
for i in range(8):
    bitplane = (image>>i) & 1
    plt.subplot(2,4,i+1)
    plt.imshow(bitplane, cmap='gray')
    plt.title("Bit plane: " + str(i))
    plt.axis('off')

# plt.show()
```
