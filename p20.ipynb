```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
image = cv2.imread("sink.jpg", 0)

# Thresholding
_, threshold = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)

#Region-based segmentation
num_labels, labels = cv2.connectedComponents(threshold)

# Create segmented image with different colors
segmented = np.uint8(255 * labels / max(num_labels - 1, 1))

# total_regions = num_labels - 1

# denominator = max(total_regions, 1)

# scaled_labels = 255 * labels / denominator

# segmented = np.uint8(scaled_labels)

# Display images
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(threshold, cmap="gray")
plt.title("Thresholding")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(segmented, cmap="nipy_spectral")
plt.title("Region Segmentation")
plt.axis("off")

plt.show()

print("Total Regions:", num_labels - 1)
```
