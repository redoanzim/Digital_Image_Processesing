```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('sink.jpg',0)
plt.imshow(img, cmap='gray')
```

```python
def convolution(image, kernel):
    (h,w)= image.shape
    result = np.zeros((h,w))
    for i in range(1,h-1):
        for j in range(1, w-1):
            region = image[i-1:i+2, j-1:j+2]
            result[i][j]=np.sum(region*kernel)
    return result
```

```python
sobel_x = np.array([
    [-1,0,1],
    [-2,0,2],
    [-1,0,1]
])

sobel_y = np.array([
    [-1,-2,-1],
    [0,0,0],
    [1,2,1]
])
```

```python
sx= convolution(img,sobel_x)
sy = convolution(img, sobel_y)
sx= np.abs(sx)
sy= np.abs(sy)
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,3,1)
plt.imshow(img, cmap='gray')
plt.axis('off')
plt.title('gray image')

plt.subplot(1,3,2)
plt.imshow(sx, cmap='gray')
plt.axis('off')
plt.title('Vertical_sobel')

plt.subplot(1,3,3)
plt.imshow(sy, cmap='gray')
plt.axis('off')
plt.title('Horizontal_sobel')

plt.show()
```
