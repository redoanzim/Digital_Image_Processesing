```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("andrew.jpg", 0)
```

```python
Gx = np.array([[-1, 0, 1],
               [-2, 0, 2],
               [-1, 0, 1]])

Gy = np.array([[-1, -2, -1],
               [ 0,  0,  0],
               [ 1,  2,  1]])
```

```python
rows, cols = img.shape
vertical = np.zeros((rows, cols))
horizontal = np.zeros((rows, cols))

padded_img = np.pad(img, 1, mode='constant')

for i in range(rows):
    for j in range(cols):
        window = padded_img[i:i+3, j:j+3]
        vertical[i, j] = np.sum(window * Gx)
        horizontal[i, j] = np.sum(window * Gy)

vertical = np.abs(vertical)
horizontal = np.abs(horizontal)
```

```python
# Show images
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(vertical, cmap="gray")
plt.title("Vertical Edges")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(horizontal, cmap="gray")
plt.title("Horizontal Edges")
plt.axis("off")

plt.show()
```
